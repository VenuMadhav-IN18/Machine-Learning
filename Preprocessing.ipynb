```python
# Which numerical feature has the largest range?

(numerical.max()-numerical.min()).idxmax()
```

```python
from numpy import quantile
# Which numerical feature appears to contain an outlier?
for col in numerical:
  q1=numerical[col].quantile(0.25)
  q3=numerical[col].quantile(0.75)
  iqr=q3-q1
  upperlimit=q3+1.5*iqr
  lowerlimit=q1-1.5*iqr
  outliers = numerical[
        (numerical[col] < lowerlimit) |
        (numerical[col] > upperlimit)
    ]
  if len(outliers)>0:
    print(col,'Contains Outliers')
```

```python
# Which categorical feature is Binary?
for col in categorical:
  if categorical[col].nunique()==2:
    print(col)
```

```python
# Which categorical feature is Ordinal?
for col in categorical:
  print(col,':',categorical[col].unique())

# "Education is Ordinal because it is Ordered"
```

```python
# Which categorical feature is Nominal?
for col in categorical:
  print(col,':',categorical[col].unique())

# "City is Nominal because it does not have any Order"
```

### **Step 9 Fit a LabelEncoder on the columns where they are required?**

Read the [documentation](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.LabelEncoder.html).

1) Why is LabelEncoder suitable for the Gender column?
- LabelEncoder is suitable because Gender is a binary categorical feature (e.g., Male/Female). Since there are only two categories, converting them into numerical values such as 0 and 1 is sufficient.
2) What do you think LabelEncoder learns during fit()?
- Finds all the unique categories.

- Assigns each category a unique integer.
3) Which attribute should contain the learned information?

1) fit the encoder

```python
# code here
# check encoder.classes_
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

encoder.fit(df["Gender"])

print(encoder.classes_)
```

2) transform the features

```python
# code here
# after that use encoder.inverse_transform
encoded = encoder.transform(df["Gender"])
print(encoded)
original = encoder.inverse_transform(encoded)
print(original)
```

3) Use fit_transform

```python
# code here
# what did you observe?
encoder = LabelEncoder()

encoded = encoder.fit_transform(df["Gender"])

print(encoded)
print(encoder.classes_)
```

- fit() learns the unique categories.
- transform() converts categories into numbers.
- fit_transform() performs both operations in one step.
- inverse_transform() converts the encoded values back to the original categories.

### **Final Step, Now that you understand every preprocessing technique individually**

1) build a single preprocessing pipeline.

2) Finally, combine them using ColumnTransformer, refer to    [documentation](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html) and transform the complete dataset.

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder
)

num_cols = ["Age", "Salary","Experience"]
ordinal_cols = ["Education"]
nominal_cols = ["City"]
binary_cols = ["Gender"]

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(categories=[
        ["High School", "Bachelor", "Master", "PhD"]
    ]))
])

nominal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

binary_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder())
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("ordinal", ordinal_pipeline, ordinal_cols),
    ("nominal", nominal_pipeline, nominal_cols),
    ("binary", binary_pipeline, binary_cols)
])

X_processed = preprocessor.fit_transform(df)

print(X_processed)
```
