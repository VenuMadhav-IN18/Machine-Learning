### Lowercase Conversion


```python
text='I LOVE Machine Learning'
text.lower()
```

### Punctuation Removal

```python
import string
text='Amazing!!!'
clean = text.translate(str.maketrans('','',string.punctuation))
print(clean)
```

### Removing Unnecessary Numbers

```python
import re
text='iPhone 16 costs 80000'
re.sub(r'\d+','',text)
```

### Remove Extra Spacees

```python
text = 'Machine     Learning'
" ".join(text.split())
```

### Remove URLS


```python
import re
text = 'Visit https://google.com'
re.sub(r'https\S+','',text)
```

### Remove HTML Tags

```python
from bs4 import BeautifulSoup
text = "<p>Hello World</p>"
BeautifulSoup(text,'html.parser').get_text()
```

### Remove Emoji

```python
import emoji
text = 'She added a smiling face emoji 😊,but she was crying 🤣🤣'
emoji.replace_emoji(text,'')
```

### Tokenization

#### Sentence Tokenization

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

```python
corpus="""
Python is amazing.
Machine Learning is Powerful.
NLP is Natural Language Processing.
"""

sent_tokenize(corpus)
```

```python
from nltk.tokenize import word_tokenize

text='This is machine learning class'

word_tokenize(text)
```

```python
from nltk.tokenize import TreebankWordTokenizer

text='This is machine learning class'

tokenizer = TreebankWordTokenizer()

tokenizer.tokenize(text)
```

### Regular Expression Tokenization

```python
from nltk.tokenize import RegexpTokenizer
tokenizer=RegexpTokenizer(r'\w+')
tokenizer.tokenize('This is machine learning class')
```

### StopWords Removal

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))
print(list(stop_words)[:20])
```

```python
from nltk.tokenize import word_tokenize
sent='This movie is really very amazing'
tokens=word_tokenize(sent)
filtered=[word for word in tokens if word.lower() not in stop_words]
filtered
```

### Stemming

```python
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies", "running"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

### Lemmatization

```python
import nltk
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "running"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos='v'))
```
