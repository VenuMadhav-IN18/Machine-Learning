```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Age':[25,23,26,28,32],
    'Salary':[45000,50000,67000,56000,90000],
    'Gender':['Male','Female','Female','Male','Female'],
    'City':['Delhi','Mumbai','Chennai','Mumbai','Hyderabad'],
    'Education':['UG','PG','PHD','PG','UG']
})
```

```python
df['Age'].std()
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x=df[['Age','Salary']]
```

```python
x
```

```python
scaler=StandardScaler()
```

```python
scaled_data=scaler.fit_transform(x)
```

```python
scaled_data
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
y=df[['Age','Salary']]
```

```python
Min_Max_scaler=MinMaxScaler()
```

```python
scaled_data=Min_Max_scaler.fit_transform(y)
```

```python
scaled_data
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder(sparse_output=False)
```

```python
encoded=encoder.fit_transform(df[['Gender']])
```

```python
encoded
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# with order
encoder=OrdinalEncoder(categories=[['UG','PG','PHD']])
encoded=encoder.fit_transform(df[['Education']])
encoded
```

```python
# without order
encoder=OrdinalEncoder()
encoded=encoder.fit_transform(df[['Education']])
encoded
```
