```python
import pandas as pd

data = {
    "Free":    ["Yes", "Yes", "Yes", "No",  "No",  "No",  "Yes", "No",  "Yes", "No"],
    "Offer":   ["Yes", "Yes", "No",  "Yes", "No",  "No",  "No",  "Yes", "Yes", "No"],
    "Money":   ["Yes", "No",  "Yes", "No",  "Yes", "No",  "Yes", "No",  "Yes", "No"],
    "Spam":    ["Yes", "Yes", "Yes", "No",  "No",  "No",  "Yes", "No",  "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Spam'].unique()
```

```python
spam_count ,not_spam_count =df['Spam'].value_counts()
```

```python
spam_prior=spam_count/len(df)
not_spam_prior=not_spam_count/len(df)
```

```python
print(spam_prior)
print(not_spam_prior)
```

```python
# Spam Dataset
spam_df=df[df['Spam']=='Yes']
```

```python
spam_df
```

```python
# Spam Likelihood

# Free
free_spam_count = (spam_df['Free'] == 'Yes').sum()
free_spam_likelihood=free_spam_count/len(spam_df)

# Offer
offer_spam_count = (spam_df['Offer'] == 'Yes').sum()
Offer_spam_likelihood=offer_spam_count/len(spam_df)

# Money
money_spam_count = (spam_df['Money'] == 'Yes').sum()
money_spam_likelihood=money_spam_count/len(spam_df)

print(free_spam_likelihood)
print(Offer_spam_likelihood)
print(money_spam_likelihood)
```

```python
not_spam_df=df[df['Spam']=='No']
print(not_spam_df)
```

```python
# Likelihood of Not Spam

free_not_spam_count=(not_spam_df['Free']=='Yes').sum()
free_not_spam_likelihood = (                           # Laplace Smoothing because of free has count 0
    (free_not_spam_count + 1)
    / (len(not_spam_df) + 2)
)

offer_not_spam_count=(not_spam_df['Offer']=='Yes').sum()
offer_not_spam_likelihood=offer_not_spam_count/len(not_spam_df)

money_not_spam_count=(not_spam_df['Money']=='Yes').sum()
money_not_spam_likelihood=money_not_spam_count/len(not_spam_df)


print(free_not_spam_likelihood)
print(offer_not_spam_likelihood)
print(money_not_spam_likelihood)
```

```python
# New Email
'''Free   = Yes
   Offer  = Yes
   Money  = No '''
```

```python
# Money = No
no_money_spam_count = (spam_df['Money'] == 'No').sum()
no_money_spam_likelihood = no_money_spam_count/len(spam_df)

no_money_spam_count = (not_spam_df['Money'] == 'No').sum()
no_money_not_spam_likelihood = no_money_spam_count/len(not_spam_df)

print(no_money_spam_likelihood)
print(no_money_not_spam_likelihood)
```

```python
# Spam Score

spam_score=spam_prior*free_spam_likelihood*Offer_spam_likelihood*no_money_spam_likelihood
spam_score
```

```python
not_spam_score=not_spam_prior*free_not_spam_likelihood*offer_not_spam_likelihood*no_money_not_spam_likelihood
not_spam_score
```

```python
if spam_score > not_spam_score:
  print('It is Spam')
else:
  print('it is not a Spam')
```
